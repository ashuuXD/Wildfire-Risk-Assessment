## Why a time-based split and not a random split?
A random split would let the model see weather patterns from dates *after* the ones it is being tested on, which never happens in a real forecasting deployment. The chronological split (earliest 70% train / next 15% validation / final 15% test) gives an honest estimate of how the model would perform when forecasting genuinely future weeks.
